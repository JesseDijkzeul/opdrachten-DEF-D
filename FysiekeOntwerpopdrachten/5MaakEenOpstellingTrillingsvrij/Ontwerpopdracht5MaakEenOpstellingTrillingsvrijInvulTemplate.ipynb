```python
g = 9.81
f_0 = 2.5
delta_x = g/((2*np.pi*f_0)**2)
print(delta_x *10**2) #m
```

## Opdracht 3: Plot de grafiek.

```python
def mass_constant(m):
    C = (m*g)/delta_x
    return C
x_test = np.linspace(0.001,10,100)
plt.plot(x_test,mass_constant(x_test), label = "Delta_x")
constante = mass_constant(x_test)/x_test
print("de verhouding van C en m is",constante[0:1])
constante_2 = 0.5*mass_constant(x_test)
plt.plot(x_test, constante_2, label = "2*Delta_x")
constante_3 = 0.2*mass_constant(x_test)
plt.plot(x_test, constante_3, label = "5*Delta_x")
plt.xlabel("massa [kg]")
plt.ylabel("veerconstante [N/m^2]")
plt.legend()
```

## Opdracht 4: Verander $\Delta x$.

We kunnen $\Delta x$ verhogen door:

- Massa te verhogen, aangezien $\Delta x = \frac{mg}{C}$
- De eigenfrequentie te verkleinen, waardoor $\omega_0$ kleiner wordt. Daaruit volgt dat $\Delta x = \frac{g}{\omega_0} $ groter wordt
 

Als groep gaan we mikken op een $\Delta x$ van: 6 cm. Dit is rendabel omdat de berekende 4 cm is. Aangezien dit een randgeval is en bijvoorbeeld niet gebruik is gemaakt van demping, is dit maar een schatting, waardoor de daadwerkelijke waarde hoger kan liggen.
